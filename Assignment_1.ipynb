```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Northgate house style (optional, but keeps your charts consistent)
NAVY = "#22303A"
RUST = "#C0562F"
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"]   = (8, 5)
plt.rcParams["axes.titlesize"]   = 13
plt.rcParams["axes.titleweight"] = "bold"
```

```python
import os
students = pd.read_csv(os.path.join(DATA_DIR, "student_records.csv"))
weekly   = pd.read_csv(os.path.join(DATA_DIR, "weekly_activity.csv"))
courses = pd.read_csv(os.path.join(DATA_DIR, "course_enrollments.csv"))
```

# Q1 · One quantitative variable
**How many hours per week do Northgate students work?**

```python
fig, ax = plt.subplots()
sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax)
ax.set_title("Hours Students Work Per Week")
ax.set_xlabel("Hours")
ax.set_ylabel("Number of Students")
plt.show()
```

S - The histogram displays the amount of hours students work per week from the students_records dataset.

T - We are looking to see if there is a trend in the amount of hours a student works.

A - Based on this chart, we can see that most students cluster between 10-20 hours  a week.

R - Something odd I noticed is that there is a slight outlier around the negatives. Possibly an error in data input as it is not possible to work negative hours.


# Q2 · One categorical variable

**How does enrollment compare across majors?**

```python
order = students["major"].value_counts().index
fig, ax = plt.subplots()
sns.countplot(data=students, y="major", order=order, color=NAVY, ax=ax)
ax.set_title("Enrollment by Major")
ax.set_xlabel("Number of students")
ax.set_ylabel("Major")
plt.show()
```

S - This bar chart displays the comparison of students enrolled in each major based from the students_records dataset.


T - We are looking to see how enrollment compares and which majors have the most and least enrollment.


A - According to the chart, Cybersecurity has the most enrollment while Info Technology at the bottom has the least.


R - However, there are multiple errors on this chart, many duplicates were made in the dataset. Cybersecurity, Data Science, Information Science, and Information Technology were repeated and have inconsistent spellings.


# Q3 · A quantity, split by category

**Does final GPA differ across majors?**

```python
fig, ax = plt.subplots()
sns.boxplot(data=students, x="final_gpa", y="major", color=NAVY, ax=ax)
ax.set_title("Final GPA Across Majors")
ax.set_xlabel("Final GPA")
ax.set_ylabel("Major")
plt.show()
```

S - This box plot displays the relation between final GPA and majors from the students_records dataset.


T - We are trying to analyze how each final GPA differs based on major and what outliers are present.


A - According to this chart, the median for each major seems to be consistent and be of range. Not one major or the other is truly lacking behind.


R - Once again, with the duplicates and inconsistent spelling from the dataset, this chart has multiple errors that can make it difficult to get an accurate comparison. There are also outliers around 0.0 that can be concerning.


# Q4 · Two quantitative variables

**Is there a relationship between how far students commute and their final GPA?**

```python
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa",
            scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
            line_kws={"color": RUST}, ax=ax)
ax.set_title("Final GPA vs. Commute Miles")
ax.set_xlabel("Commute Miles")
ax.set_ylabel("Final GPA")
plt.show()
```

S - This scatter plot displays both the information on the commute distance and their final GPA from the students_records dataset.


T - We are looking to find if final GPA and commute miles are in correlation to one another.


A - According to the scatter plot, there seems to be a downward trend with the more miles commuted, the lower the final GPA is.  


R - Based on the information represented on the chart, there is not enough information to say a longer commute equals lower grades. There may be other factors such as busy work and scheduling.


# Q5 · Over time

**How does student engagement change across the semester?**

```python
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()
fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active",
             color=NAVY, marker="o", ax=ax)
ax.set_title("Average Active Minutes per Student, by Week")
ax.set_xlabel("Week of semester")
ax.set_ylabel("Avg. Minutes Active")
plt.show()
```

S - This line chart represents the average active minutes a student engages across the weeks of a semester from the weekly_activity dataset.


T - We are looking to identify if there is a trend with the engagement and the stage of the semester (week).


A - Based on the line chart, there seems to be a drop in engagement towards week 9 which correlates around the timing of Thanksgiving break.


R - As the engagement drops at week 9, it aligns with a planned academic break. We can come to infer that sudden drops are based on breaks rather than a loss of motivation.


# Part B · Two quantitative variables

**Is there a relationship between the amount of advisor meetings a student takes and their final GPA?**

```python
fig, ax = plt.subplots()
sns.regplot(data=students, x="advisor_meetings", y="final_gpa",
            scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
            line_kws={"color": RUST}, ax=ax)
ax.set_title("Final GPA vs. Advisor Meetings")
ax.set_xlabel("Advisor Meetings")
ax.set_ylabel("Final GPA")
plt.show()
```

S - This scatter plot displays the relationship between advisor meetings a student participates in and their final GPA based on the students_records dataset.


T - We are seeking to identify if there is a relationship with the amount of advisor meetings a student attends and their final GPA.


A - Based on the trend line, there is an upward trend. The more advisor meetings a student has, the higher their final GPA is.


R - However, the points on the plot seem to be consistent and students can have both a low or high final GPA no matter the amount of advisor meetings. The association does not equal causation.
